```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## Partie A : Classification (filtre anti-spam)

**1. Matrice de confusion** (classe positive = spam)

| | Prédit spam | Prédit normal |
|---|---|---|
| **Réel spam** | TP = 40 | FN = 5 |
| **Réel normal** | FP = 10 | TN = 45 |

**2 à 6.** Formules et vérification :

```python
TP, FP, FN, TN = 40, 10, 5, 45

accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
rappel = TP / (TP + FN)
f1 = 2 * precision * rappel / (precision + rappel)
specificite = TN / (TN + FP)

print("Accuracy    :", round(accuracy, 3))
print("Précision   :", round(precision, 3))
print("Rappel      :", round(rappel, 3))
print("F1-score    :", round(f1, 3))
print("Spécificité :", round(specificite, 3))
```

**7.** Filtre paresseux (tout est classé spam) : TP = 45, FP = 55, FN = 0, TN = 0

```python
TP, FP, FN, TN = 45, 55, 0, 0
print("Accuracy  :", (TP + TN) / 100)
print("Précision :", TP / (TP + FP))
print("Rappel    :", TP / (TP + FN))
```

Rappel parfait (1,0) mais précision et accuracy de 0,45 : tous les e-mails normaux sont envoyés en spam. Un bon rappel ne suffit pas, il faut le lire avec la précision.

## Partie B : Régression

`y_réel = [3, 5, 2.5, 7]` et `y_prédit = [2.5, 5, 4, 8]`

**8 à 11.** Erreurs, MAE, MSE, RMSE, R² et R² ajusté (n = 4, p = 1)

```python
y_reel = np.array([3, 5, 2.5, 7])
y_pred = np.array([2.5, 5, 4, 8])

e = y_pred - y_reel
mae = np.mean(np.abs(e))
mse = np.mean(e ** 2)
rmse = np.sqrt(mse)

ss_res = np.sum((y_reel - y_pred) ** 2)
ss_tot = np.sum((y_reel - y_reel.mean()) ** 2)
r2 = 1 - ss_res / ss_tot
n, p = 4, 1
r2_adj = 1 - (1 - r2) * (n - 1) / (n - p - 1)

print("Erreurs :", e)
print("MAE =", round(mae, 3), " MSE =", round(mse, 3), " RMSE =", round(rmse, 3))
print("R2 =", round(r2, 3), " R2 ajusté =", round(r2_adj, 3))
```

**12.** Avec une valeur aberrante (dernière prédiction = 17)

```python
e2 = np.array([2.5, 5, 4, 17]) - y_reel
print("MAE  :", round(mae, 3), "->", round(np.mean(np.abs(e2)), 3))
print("RMSE :", round(rmse, 3), "->", round(np.sqrt(np.mean(e2 ** 2)), 3))
```

La **RMSE** réagit le plus (×5 contre ×4 pour la MAE) car les erreurs sont élevées au carré.

**13.** Vérification avec scikit-learn

```python
print("MAE  :", mean_absolute_error(y_reel, y_pred))
print("MSE  :", mean_squared_error(y_reel, y_pred))
print("RMSE :", round(np.sqrt(mean_squared_error(y_reel, y_pred)), 3))
print("R2   :", round(r2_score(y_reel, y_pred), 3))
```
