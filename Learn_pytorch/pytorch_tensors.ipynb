```python
import torch
import numpy as np
#let's learn about tensor
t = torch.tensor([1,2,3,4,5,6])
print(t)
```

```python
t1 = torch.ones(3,3,dtype=torch.int)
t2 = torch.zeros(3,4)
t3 = torch.rand(3,4)
#t31 = torch.randint()
#t32 = torch.randint()
t33 = torch.rand_like(t2) #same shape
t4 = torch.empty(3,4)
t5 = torch.full((3,4),7,dtype=torch.int)
t6 = torch.arange(0,10,2)
t7 = torch.linspace(0,78,4)
t8 = torch.zeros_like(t7) #same shape
t81 = torch.ones_like(t7)
t82 = torch.empty_like(t7)
print('t1:',t1)
print('t2:',t2)
print('t3:',t3)
print('t33:',t33)
print('t4:',t4)
print('t5:',t5)
print('t6:',t6)
print('t7:',t7)
print('t8:',t8)
print('t81:',t81)
print('t82:',t82)
```
