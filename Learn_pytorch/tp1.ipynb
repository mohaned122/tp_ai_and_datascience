```python
import torch

x = torch.tensor([2.0,3.0], requires_grad=True)
y = torch.tensor([1.0,6.0], requires_grad=True)

f = x**2 + 3*y + 2

f.backward(gradient=torch.tensor([3,3]))
print(x.grad)
```

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2

y.backward(retain_graph=True)
print(x.grad)

x.grad = None

y.backward()
print(x.grad)
```
