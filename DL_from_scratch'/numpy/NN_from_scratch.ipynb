```python
import numpy as np

X = np.array([
    [2, 7],
    [5, 8],
    [8, 6],
    [1, 4]
])

y = np.array([
    [1],
    [0],
    [1],
    [0]
])
m = X.shape[0]
print(X.shape)
```

```python
np.random.seed(42)

W1 = np.random.randn(2, 3)
b1 = np.zeros((1, 3))

W2 = np.random.randn(3, 1)
b2 = np.zeros((1, 1))
```

```python
print(W1.shape)  # (2, 3)
print(b1.shape)  # (1, 3)

print(W2.shape)  # (3, 1)
print(b2.shape)  # (1, 1)
```

```python
Z1 = np.dot(X, W1) + b1

print(Z1)
print(Z1.shape)
```

```python
def sigmoid(x: np.ndarray) -> np.ndarray:
    return 1 / (1 + np.exp(-x))
```

```python
# Forward propagation
A1 = sigmoid(Z1)
```

```python
Z2 = A1 @ W2 + b2
A2 = sigmoid(Z2)

print("Z1:", Z1.shape)
print("A1:", A1.shape)
print("Z2:", Z2.shape)
print("A2:", A2.shape)

print("Predictions:")
print(A2)
```

```python
predicted_classes = (A2 >= 0.5).astype(int)

accuracy = np.mean(predicted_classes == y)

print("Predictions:", predicted_classes)
print("Accuracy:", accuracy)
```

```python
def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-15

    y_pred = np.clip(
        y_pred,
        epsilon,
        1 - epsilon
    )

    loss = -np.mean(
        y_true * np.log(y_pred)
        + (1 - y_true) * np.log(1 - y_pred)
    )

    return loss
```

```python
def forward(X):
    Z1 = X @ W1 + b1
    A1 = sigmoid(Z1)

    Z2 = A1 @ W2 + b2
    A2 = sigmoid(Z2)

    return Z1, A1, Z2, A2


def backward(X, y, A1, A2):
    global W1, W2

    # Output layer gradients
    dZ2 = A2 - y
    dW2 = A1.T @ dZ2 / m
    db2 = np.sum(dZ2, axis=0, keepdims=True) / m

    # Hidden layer gradients
    dZ1 = (dZ2 @ W2.T) * A1 * (1 - A1)
    dW1 = X.T @ dZ1 / m
    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    return dW1, db1, dW2, db2
```

```python
print(binary_cross_entropy(predicted_classes, y))
```

```python
# Training
learning_rate = 0.1
epochs = 10000

for epoch in range(epochs):

    Z1, A1, Z2, A2 = forward(X)

    loss = binary_cross_entropy(y, A2)

    dW1, db1, dW2, db2 = backward(
        X, y, A1, A2
    )

    # Gradient descent
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    if epoch % 1000 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.4f}")


# Final predictions
_, _, _, predictions = forward(X)

print("\nProbabilities:")
print(predictions)

print("\nPredicted classes:")
print((predictions >= 0.5).astype(int))
```
