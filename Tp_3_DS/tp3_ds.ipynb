## II. Transformer les données avec map et apply

```python
#montant = prix_unitaire × quantite

df['manttant'] = df['prix_unitaire'] * df['quantite']
```

```python
df['sexe_std'].unique()
```

```python
df['sexe_std'] = df['sexe'].map({
    'Homme':'M',
    'Femme':'F',
    'M':'M',
    'F':'F'
})
```

```python
#Avec map() et une fonction lambda, créer la colonne prenom contenant le premier mot de la colonne client.
df['prenom'] = df['client'].map(lambda x: x.split(' ')[0])
```

Écrire une fonction gamme(prix) qui renvoie « Économique » si le prix est < 100, « Standard » s’il
est < 500, et « Premium » sinon. L’appliquer avec apply() pour créer la colonne gamme. Afficher
les effectifs de chaque gamme.

```python
def  gamme(prix):
    if prix < 100:
        return 'economique'
    elif prix < 500:
        return 'standare'
    else:
        return 'premuim'

df['gamma']=df.apply(lambda row: gamme(row['prix_unitaire']), axis=1)
```

5. Convertir date_cmd et date_livraison en dates avec pd.to_datetime(). Écrire une fonction
statut(row) qui calcule le délai de livraison en jours (.days) et renvoie « Incohérent » si le délai
est négatif, « Express » s’il est ≤ 3 jours, « Normal » s’il est ≤ 5 jours et « Lent » sinon. L’appliquer
ligne par ligne (axis=1) pour créer la colonne statut.

```python
df['date_cmd'] = pd.to_datetime(df['date_cmd'])
df['date_livraison'] = pd.to_datetime(df['date_livraison'])
df['delai'] = df['date_livraison'] - df['date_cmd']
```

```python
def status(row):
    delai = (row['date_livraison'] - row['date_cmd']).days
    if delai < 0:
        return 'incoherente'
    elif delai < 3:
        return 'express'
    elif delai < 5:
        return 'normal'
    else:
        return 'post tunisien'
```

```python
df['status_liv']=df.apply(lambda row: status(row), axis=1)
```
