# Exercice 1 — Base de données cancer du sein

### 1. Charger la base de données dans la variable `breast`

`load_breast_cancer` (de `sklearn.datasets`) charge un jeu de données sur le cancer du sein.

```python
from sklearn.datasets import load_breast_cancer

breast = load_breast_cancer()
```

```python
print(breast.target_names)

T = breast.target
print("\nPremières valeurs de target :", T[:20])

for i in [0, 1]:
    print("la classe: %s, contient un nombre d'exemplaires: %s" % (i, len(T[T == i])))
```

### 8. Créer un fichier CSV `liste1.csv` et le lire


```python
import csv

notes = [
    ['ID', 'Noms', 'Anglais', 'Maths', 'Arabe', 'Allemend'],
    ['1', 'Sarra', '15', '17', '10', '20'],
    ['2', 'Olfa', '10', '9', '17', '15'],
    ['3', 'Ahmed', '11', '9', '6', '20'],
    ['4', 'Taheni', '10', '5', '18', '19'],
    ['5', 'Rami', '19', '18', '16', '19'],
]

# Écriture du fichier ('w' = écriture, efface le contenu existant)
with open('liste1.csv', 'w', newline='') as f:
    ecriture = csv.writer(f)
    for ligne in notes:
        ecriture.writerow(ligne)
print("Fichier liste1.csv créé.")
```

```python
# Lecture du fichier
with open('liste1.csv', newline='') as g:
    T = []
    lecture = csv.reader(g)
    print('les lignes sont:')
    for l in lecture:        # itération ligne par ligne
        print(l)             # afficher la ligne
        T.append(l)          # ajouter la ligne à la liste T
```

```python
print(T)
```

### 9. Reconstruire le tableau dans un fichier `liste2.csv`


```python
with open('liste2.csv', 'w', newline='') as h:
    ecriture = csv.writer(h)          # préparation à l'écriture
    for j in T:                       # pour chaque ligne du tableau
        ecriture.writerow(j)          # écrire cette ligne

print('longueur de T :', len(T))
```

```python
# Vérification : liste2.csv contient le même tableau
with open('liste2.csv', newline='') as f:
    lecture = csv.reader(f)
    for k in lecture:
        print(k)
```

### 10. Ajouter une ligne à `liste1.csv`


```python
with open('liste1.csv', 'a', newline='') as g:
    ecriture = csv.writer(g)
    ecriture.writerow(['6', 'Asma', '11', '3', '19', '14'])
```

```python
with open('liste1.csv', newline='') as f:
    lecture = csv.reader(f)
    for k in lecture:
        print(k)
```

# Exercice 2 — Films sur une plateforme de streaming (Pandas)

### 1. Création du DataFrame `films`

```python
import pandas as pd

films = pd.DataFrame({
    'titre': ['Film A', 'Film B', 'Film C', 'Film D'],
    'realisateur': ['Réalisateur 1', 'Réalisateur 2', 'Réalisateur 3', 'Réalisateur 4'],
    'note': [4.2, 3.7, 4.8, 4.0],
    'duree': [120, 95, 140, 110],
    'pays': ['France', 'Canada', 'France', 'Espagne'],
    'genre': ['Drame', 'Comédie', 'Action', 'Drame']
})
```

### 2. Affichage du DataFrame

```python
print(films)
```

### 3. Sauvegarde dans `films_streaming.csv` (sans l'index)

```python
films.to_csv('films_streaming.csv', index=False)
```

### 4. Message de vérification

```python
print("Le fichier 'films_streaming.csv' a été créé avec succès !")
```

### 5. Importation dans `films_importes`

```python
films_importes = pd.read_csv('films_streaming.csv')
print(films_importes)
```

### 6. Médiane de la colonne `note`

```python
print("Médiane des notes :", films['note'].median())
```

### 7. Pays représentés (sans doublons)

```python
print(films['pays'].unique())
```

### 8. Nombre de films par pays : Series `films_par_pays`

```python
films_par_pays = films['pays'].value_counts()
print(films_par_pays)
```

### 9. Ajout de la colonne `prix`

```python
films['prix'] = [12.5, 9.9, 15.0, 11.5]
print(films)
```

```python
films.loc[len(films)] = ['Film E', 'Réalisateur 5', 4.5, 125, 'Canada', 'Action', 13.5]
```

### 12. Affichage du DataFrame mis à jour

```python
print(films)
```

### 13. Médiane des prix

```python
print("Médiane des prix :", films['prix'].median())
```

### 14. Moyenne des prix

```python
print("Moyenne des prix :", films['prix'].mean())
```

### 15. Nombre de films dont la note est ≥ 4.2
